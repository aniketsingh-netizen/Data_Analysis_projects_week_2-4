```python
import pandas as pd
filename = 'SQL_Sales_Dataset_200_Rows.xlsx - Sheet1.csv'
df = pd.read_csv(filename)

print("--- First 5 Rows ---")
print(df.head())

print("\n--- Basic Information ---")
print(df.info())

print("\n--- Summary Statistics ---")
print(df.describe())
df = df.drop_duplicates()
df = df.fillna(0)
print("\n--- Missing values and duplicates handled! ---")
if 'category' in df.columns and 'total_price' in df.columns:
    revenue_by_category = df.groupby('category')['total_price'].sum()
    print("\n--- Total Revenue by Category ---")
    print(revenue_by_category)
if 'region' in df.columns and 'total_price' in df.columns:
    sorted_df = df.sort_values(by=['region', 'total_price'], ascending=[True, False])
    print("\n--- Sorted Data Preview ---")
    print(sorted_df.head())
numerical_df = df.select_dtypes(include=['number'])
correlation_matrix = numerical_df.corr()
print("\n--- Correlation Matrix ---")
print(correlation_matrix)
```
