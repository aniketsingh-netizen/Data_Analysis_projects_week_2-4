```python
import pandas as pd
import numpy as np
file_path = '/content/Sample data (1).xlsx - Sheet1.csv'
df = pd.read_csv(file_path)

print("--- 1. DATASET OVERVIEW ---")
print(f"Total Rows: {df.shape[0]} | Total Columns: {df.shape[1]}")
print("\nFirst 5 rows of the dataset:")
display(df.head())

print("\n--- 2. DATA TYPES & COLUMNS ---")
print(df.info())
print("\n--- 3. MISSING VALUES BEFORE CLEANING ---")
missing_summary = df.isnull().sum()
print(missing_summary[missing_summary > 0])
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna('Unknown')
    else:
        df[col] = df[col].fillna(df[col].median())
duplicate_count = df.duplicated().sum()
print(f"\n--- 4. DUPLICATES FOUND: {duplicate_count} ---")
if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows have been successfully removed.")
print("\n--- 5. KEY STATISTICAL METRICS ---")
display(df.describe(include='all'))
cleaned_file_name = "cleaned_capstone_data.csv"
df.to_csv(cleaned_file_name, index=False)
print(f"\n--- SUCCESS: Cleaned dataset saved as '{cleaned_file_name}' ---")
```
